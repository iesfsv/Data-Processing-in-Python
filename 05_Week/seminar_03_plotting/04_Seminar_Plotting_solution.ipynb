### Part 1 - Matplotlib basics

**TASK**: Import the macro dataset using pandas. Hint: the file is in Excel format, so use the `read_excel` function.

```python
macroData = pd.read_excel("data/macroData1.xlsx")
```

**TASK**: Explore the dataset. Print the dataset and the data types of its columns.

```python
print(macroData)
print(macroData.dtypes)
```

**TASK**: Set `time` as the index and print only the `RGDP` column. Hint: use the `set_index` method.

```python
macroData = macroData.set_index("time")
print(macroData["RGDP"])
```

**TASK**: Let's visualize `RGDP` using matplotlib. Hint: create the figure with `fig, ax = plt.subplots()` and use the `ax.plot` method.

```python
fig, ax = plt.subplots()
ax.plot(macroData.index, macroData["RGDP"], label="RGDP")
ax.set_ylabel("Millions of dollars")
ax.set_xlabel("Time")
ax.legend()
ax.set_title("Plot of real gross domestic product over time")
plt.show()
```

**QUESTION**: What would happen if we did not set `time` as the index?

**TASK**: Now plot RGDP and PGDP (potential GDP) in the same graph and include both in the `legend`. Hint: specify `label` in each `plot` call and use the `legend` method.

```python
fig, ax = plt.subplots()
ax.plot(macroData.index, macroData["RGDP"], label="RGDP")
ax.plot(macroData.index, macroData["PGDP"], label="PGDP", color="green")
ax.set_ylabel("Millions of dollars")
ax.set_xlabel("Time")
ax.legend()
plt.show()
```

**TASK**: Plot CPI alongside the other two variables. Will it work? Hint: you may need a secondary y-axis (`twinx`).

```python
fig, ax = plt.subplots()
ax.plot(macroData.index, macroData["RGDP"], label="RGDP")
ax.plot(macroData.index, macroData["PGDP"], label="PGDP", color="orange")
ax2 = ax.twinx()
ax2.plot(macroData.index, macroData["CPI"], label="CPI", color="green")
ax2.set_ylabel("%")
ax.set_ylabel("Millions of dollars")
ax.set_xlabel("Time")
ax.legend(loc="upper left")
ax2.legend(loc="upper right")
plt.show()
```

**TASK**: Calculate the output gap, `(RGDP - PGDP) / RGDP * 100`, in percent and plot it as a bar chart for 2015-2025. Hint: filter the rows using the index and use the `ax.bar` method.

```python
macroData["OG"] = (macroData["RGDP"] - macroData["PGDP"]) / macroData["RGDP"] * 100
macroDataSamp = macroData[macroData.index > "2015-01-01"]

fig, ax = plt.subplots()
ax.bar(macroDataSamp.index, macroDataSamp["OG"], label="Output Gap", width=50)
ax.set_ylabel("% of RGDP")
ax.set_xlabel("Time")
ax.legend()
ax.set_title("Output Gap over Time")
plt.show()
```

**TASK**: Plot inflation (CPI) as a histogram. Hint: use the `ax.hist` method.

```python
fig, ax = plt.subplots()
ax.hist(macroData["CPI"], label="Inflation (CPI)", bins=20, edgecolor="black")
ax.set_ylabel("Frequency")
ax.set_xlabel("% of inflation (CPI)")
ax.legend()
plt.show()
```

**TASK**: Does the mean inflation match the Fed's 2% target? Check it. Hint: use the `ax.axvline` method to add the target and the mean to the histogram.

```python
fig, ax = plt.subplots()
ax.hist(macroData["CPI"], label="Inflation (CPI)", bins=20, edgecolor="black", color="lightblue")
ax.set_ylabel("Frequency")
ax.set_xlabel("% of inflation (CPI)")
ax.axvline(2, label="Fed target", color="red", linestyle="--")
ax.axvline(macroData["CPI"].mean(), label="Mean of CPI", linestyle="-.", color="green")
ax.legend()
plt.show()
```

**TASK**: Now plot the gap of CPI and UNR (CPI target is 2% and UNR = 4.2%). Hint: use `ax.axhline()` method

```python
fig, ax = plt.subplots()
ax.plot(macroData.index, macroData["CPI"] - 2, label="CPI Gap")
ax.plot(macroData.index, macroData["UNR"] - 4.2, label="UNR Gap")
ax.set_xlabel("Time")
ax.set_ylabel("% of Gap")
ax.axhline(0, label="Target", color="red", linestyle="--")
ax.legend()
plt.show()
```

**TASK**: Plot histograms of these 4 random variables with different distributions. Hint: use `plt.subplots(2, 2)` function

```python
n1, n2, n3, n4 = np.random.randint(100, 501, size=4)

x1 = np.random.normal(0, 1, n1)
x2 = np.random.uniform(0, 1, n2)
x3 = np.random.exponential(1, n3)
x4 = np.random.beta(2, 5, n4)
```

```python
fig, axes = plt.subplots(2, 2)
axes[0, 0].hist(x1)
axes[0, 0].set_title("Normal")
axes[0, 1].hist(x2)
axes[0, 1].set_title("Uniform")
axes[1, 0].hist(x3)
axes[1, 0].set_title("Exponential")
axes[1, 1].hist(x4)
axes[1, 1].set_title("Beta")
fig.tight_layout()
plt.show()
```

### Part 2 - Seaborn

Seaborn is built on top of matplotlib. It needs less code and has nicer defaults. It draws on the same `ax`, so we just pass `ax=ax` and can still use `ax.set_title`, `ax.axhline` and so on afterwards.

**TASK**: Load the seaborn dataset `penguins`, drop the rows with missing values and check the data types of the columns. Hint: use the `load_dataset` function and the `dropna` method.

```python
pengData = sns.load_dataset("penguins").dropna()
print(pengData.dtypes)
```

**TASK**: Make a bar plot of the body mass of penguins by species. Hint: use the `barplot` function from seaborn and pass `ax=ax`.

```python
fig, ax = plt.subplots()
sns.barplot(pengData, x="species", y="body_mass_g", ax=ax)
ax.set_title("Body mass of penguins")
plt.show()
```

**TASK**: Now load the `flights` dataset from seaborn and check its columns.

```python
flightData = sns.load_dataset("flights")
print(flightData.dtypes)
```

**TASK**: Pivot the table so that the index is `month`, the columns are `year` and the values are `passengers`. Hint: use the `pivot` method.

```python
flightPivot = flightData.pivot(index="month", columns="year", values="passengers")
print(flightPivot)
```

**TASK**: Create a heatmap from the pivoted flights data. Hint: use the `heatmap` function from seaborn.

```python
fig, ax = plt.subplots()
sns.heatmap(flightPivot, annot=True, fmt="d", ax=ax)
ax.set_title("Heatmap Plot")
plt.show()
```
