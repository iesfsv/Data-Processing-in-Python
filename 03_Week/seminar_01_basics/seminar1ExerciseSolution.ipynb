#### c. Rewrite the sum functions from the previous exercise so that they sum numbers and characters separately and work with a mixed list

```python
mixedList = [0, 'o', 1, 's', 'c', 5, "a", 8, 9, "r"]
```

Write the function in the code cell below

```python
def sumAll(x: list):
    sumNum = 0
    sumChar = ""

    for i in x:
        if type(i) == int:
            sumNum += i
        else:
            sumChar = sumChar + i
    return sumNum, sumChar
```

Use the function in the code cell below

```python
totalNum, totalChar = sumAll(mixedList)
print(totalNum)
print(totalChar)
```

#### Let's sum only the numbers from the mixed list, but using `exceptions`

```python
mixedList = [0, 'o', 1, 's', 'c', 5, "a", 8, 9, "r"]
```

```python
def sumException(x: list):
    total = 0
    for i in x:
        try:
            total += i
        except TypeError as e:
            print("Exception. Reason:", e)
    print(f"Final summation is {total}")
```

## Exercise 6: Unique Elements Finder

Write a function `uniqueElements(lst)` that takes a list and returns a new list with only the unique elements from the original list (i.e., removing duplicates).
You can use a loop or a list comprehension for this exercise.

```python
letters = ['a', 'b', 'f', 'a', 'c', 'd', 'c', 'f', 'b']
```

Write the function in the code cell below

```python
def uniqueElements(lst: list):
    result = []
    for item in lst:
        if item not in result:
            result.append(item)
    return result
```

Test the function in the code cell below and sort the result

```python
sortedUnique = sorted(uniqueElements(letters))
print(sortedUnique)
```

## Exercise 7: Text Analyzer

Write a function `analyzeText` that takes a string and returns:
1. Number of words.
2. Number of vowels.
3. Whether the string contains the word 'Python' (case insensitive).

```python
exampleString = "Python is amazing"
vowels = "aeiouAEIOU"
```

Write the function

```python
def analyzeText(text: str):
    wordCount = len(text.split())
    vowelCount = sum(1 for char in text if char in vowels)
    containsPython = 'python' in text.lower()
    return wordCount, vowelCount, containsPython
```

Use the function in the code cell below

```python
wordCount, vowelCount, containsPython = analyzeText(exampleString)
print(f"Word count is: {wordCount}")
print(f"Vowel count is: {vowelCount}")
print(f"Does the text contain Python? {containsPython}")
```

## Exercise 8: Prime Number Checker

Write a function `isPrime(n)` that returns `True` if the number `n` is prime and `False` otherwise.

```python
def isPrime(n: int):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
```

Test the function in the code cell below

```python
print(isPrime(4))
print(isPrime(7))
```

**TASK** Optional: Write a function that returns all prime numbers between 1 and 100.

```python
def getPrimes(numbers):
    primes = [n for n in numbers if isPrime(n)]
    print(f"In total there are {len(primes)} prime numbers")
    return primes


getPrimes(range(1, 101))
```

### Exercise 9: List Manipulation
Write a function `listOperations(lst)` that returns a dictionary with:
- 'min': the smallest number
- 'max': the largest number
- 'sum': the sum of numbers
- 'average': the average of numbers (rounded to 2 decimals)

```python
exampleList = [1, 10, 30, 50, -3, 42, -50]
```

```python
def listOperations(lst: list):
    minVal = min(lst)
    maxVal = max(lst)
    sumVal = sum(lst)
    avgVal = round(sumVal / len(lst), 2)
    return {"min": minVal,
            "max": maxVal,
            "sum": sumVal,
            "average": avgVal}
```

```python
listOperations(exampleList)
```

### Exercise 10: Factorial using Recursion
Write a recursive function `factorial(n)` that returns the factorial of a number `n`.
For example, `factorial(5)` should return `120` because `5! = 5 * 4 * 3 * 2 * 1 = 120`.

```python
def factorial(n: int):
    if n <= 1:
        return 1
    return n * factorial(n - 1)
```
