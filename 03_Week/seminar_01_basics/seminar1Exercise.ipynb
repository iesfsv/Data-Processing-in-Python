### Recap: file IO

- We can load many things (data - text, numbers, images, video, etc.) into our programs.

```python
# The built-in function open() is used to open a file and return a corresponding file object.
# The open() function takes two parameters: filename and mode.
# There are four different modes for opening a file: "r", "a", "w", "x"
# "r" - Read - Default value. Opens a file for reading, error if the file does not exist
# "a" - Append - Opens a file for appending, creates the file if it does not exist
# "w" - Write - Opens a file for writing, creates the file if it does not exist
# "x" - Create - Creates the file, error if the file already exists
```

```python
f = open('newfile.txt', 'w')   # Open 'newfile.txt' for writing
f.write('Testing\n')           # Here '\n' means new line
f.write('Testing again')
f.close()
```
